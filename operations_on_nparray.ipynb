```python
import numpy as np
```

```python
# Operations on np array
"""
1.Reshaping the array
2.Indexing (1D&2D Array)
3.Fancy & Boolean Indexing
4.Slicing
"""
```

```python
# Reshaping the array
arr=np.array([[1,2,3],[4,5,6]],dtype="int64")
print(arr,arr.shape)
reshape_arr=arr.reshape((3,2))
print(reshape_arr,reshape_arr.shape)
# The reshaping operation only works when the row and column should be equal to the number of elements in the array...
# Flatten an array converting 2d array to 1d array
flattened_arr=arr.flatten()
print(flattened_arr,flattened_arr.shape)
```
