```python
"""
Integers : int32,int64
Floating point nums: float32,float64
Boolean: bool
Complex nums: complex64,complex128
String s(byte-str)&U(unicode-str)
Object:generic python objects-object
"""
```

```python
import numpy as np
```

```python
arr=np.array([1,2,3,4])
print(arr,arr.dtype)
```

```python
arr=np.array([1.,2.,3.,4.])
print(arr,arr.dtype)
```

```python
arr=np.array([True,False,True,False])
print(arr,arr.dtype)
```

```python
arr=np.array([1+2j])
print(arr,arr.dtype)
```

```python
# It is not recommended to store strings and objects in numpy
```

```python
arr=np.array([1,"Prime",{1,2,3}])
```

```python
print(arr,arr.dtype)
```
