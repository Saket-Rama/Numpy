```python
# Normalisation  - Transforming the values to make the elements fit in a specific range
```

```python
# In order to normalize the vector [[1,2],[3,4]]
"""
Formula = (arr-mean)/std_dev
std_dev=sqrt(sum((val[i]-mean)**2)/N) 
N = Number of Values
"""
```

```python
import numpy as np
```

```python
arr=np.array([[1,2],[3,4]])
mean_arr=np.mean(arr)
std_arr=np.std(arr)
```

```python
normalized_arr=(arr-mean_arr)/std_arr
print(normalized_arr)
```

```python
# Mean of the Normalized arr is always zero...
```
