```python
import numpy as np
```

```python
# Vectorisation - Peforming calculations on an entire array at once without any iteration...
arr=np.array([1,2,3,4,5])
print(arr+10)
print(arr**10)
arr2=np.array([1,2,3,4,5])
print(arr+arr2)
```

```python
arr3=np.array([1,2])
# print(arr3+arr2)
# In this example we can see the different np arrays shapes can't be added
```

```python
# Broadcasting - Allows arithematic operations of two different shape arrays
arr=np.array([1,2,3,4,5])
print(arr+5)
# Here the broadcasting concept is used because as u can see the shapes are different
arr_1d=np.array([1,2,3,4]) # 1x5
arr_2d=np.array([[1,2,3,4],[1,2,3,4]]) # 2x5
print(arr_1d+arr_2d)
# Broadcasting two rules - 
"""
Two dimensions are compatible if:

They are equal, or

One of them is 1

If neither of these conditions is met, NumPy will raise a ValueError.
"""
```
