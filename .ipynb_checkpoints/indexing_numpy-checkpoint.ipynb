```python
import numpy as np
```

```python
arr=np.array([1,2,3,4])
```

```python
# Normal indexing
print(arr[0])
# This is for 1d array
# For 2d Array
arr2=np.array([[1,2,3],[4,5,6]])
print(arr2[0][2]) # specify row and column in more than 1d array...
```

```python
# fancy indexing
idx=[0,2]
print(arr[idx])
```

```python
# Boolean Indexing
arr=np.array([1,2,3,4,5])
print(arr[arr>2])
print(arr[arr%2])
print(arr[arr%2!=0])
```
