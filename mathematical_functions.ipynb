```python
import numpy as np
```

```python
# mathematical operations...
arr=np.array([1,2,3,4,5])
print(np.sum(arr)) # sum of the numbers
print(np.prod(arr)) # Product of the numbers
print(np.max(arr)) # Max among the numbers
print(np.min(arr)) # Min among the numbers
print(np.argmin(arr)) # Index of the min value
print(np.argmax(arr)) # Index of the max value
print(np.mean(arr)) # mean of the numbers
print(np.median(arr)) # median of the numbers
print(np.std(arr)) # standard deviation of the numbers
print(np.var(arr)) # variance of the numbers
```

```python
# Power Functions
arr=np.array([1,2,3,4,5])
print(np.square(arr)) # Square of the arr
print(np.sqrt(arr)) # Square root of the arr
print(np.pow(arr,3)) # pow(number,itspower)
```

```python
# Log Functions
arr=np.array([1,2,3,4,5])
print(np.log(arr))
print(np.log10(arr))
print(np.exp(arr))
```

```python
# Rounding off functions
print(np.round(2.2511))
print(np.ceil(3.14)) # Rounding up to the higher decimal
print(np.floor(3.14)) # Rounding up to the lower decimal
print(np.trunc(2.51515)) # Removing after decimal numbers

arr=np.array([1,3,5,2,1,5,3,4])
print(np.unique(arr)) # Same as Set 
print(np.sort(arr))
arr2=np.array([-1,2,-3,4,5,-6])
print(np.abs(arr2))
```
